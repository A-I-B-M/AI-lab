```python
# -------- SET --------
s = set([1, 2, 2, 3])      # {1, 2, 3}
s.add(5)
s.remove(2)
print(s)                  # {1, 3, 5}

# -------- MAP --------
nums = [1, 2, 3, 4]
sq = list(map(lambda x: x*x, nums))
print(sq)                 # [1, 4, 9, 16]

# -------- PAIR (key:value) --------
pairs = input("(key:value): ").split()
d = {k: v for k, v in (p.split(":") for p in pairs)}
print(d)                  # {'name':'Bappy', ...}

# -------- LAMBDA --------
add = lambda a, b: a + b
print(add(3, 4))          # 7
```
