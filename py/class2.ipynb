```python
d = {
    "name": "Bappy",
    "age": 22,
    "dept": "CSE"
}

# ----- ACCESS VALUES -----
print(d["name"])            # direct access
print(d.get("age"))         # safe access (no error)

# Missing key
print(d.get("joined"))      # None

# ----- CHECK KEY EXISTS -----
print("dept" in d)          # True
print("fariha" in d)        # False

# ----- ADD NEW KEY -----
d["joined"] = "2024"
print(d)

# ----- UPDATE VALUE -----
d["age"] = 23
print(d)

# ----- REMOVE KEY -----
d.pop("dept")
print(d)

# ----- GET ALL KEYS -----
print(d.keys())             # dict_keys(['name','age','joined'])

# ----- GET ALL VALUES -----
print(d.values())           # dict_values([...])

# ----- GET ALL ITEMS (pairs) -----
print(d.items())            # dict_items([('name','Bappy'), ...])

# ----- LOOP THROUGH DICTIONARY -----
for key, value in d.items():
    print(key, ":", value)

# ----- DICTIONARY FROM INPUT -----
# Example input: name:Bappy age:22 city:Dhaka
pairs = input("(key:value): ").split()
user_dict = {k: v for k, v in (p.split(":") for p in pairs)}
print("User Dictionary:", user_dict)


```

```python
import matplotlib.patches as patches

# ----- 1. LINE PLOT -----
x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.figure(figsize=(6,6))
plt.plot(x, y, label="Sine Wave", color='blue')
plt.title("Line Plot")
plt.legend()
plt.show()


# ----- 2. SCATTER PLOT -----
x = np.random.rand(20)
y = np.random.rand(20)
sizes = np.random.randint(50, 300, size=20)
colors = np.random.rand(20)

plt.figure(figsize=(6,6))
plt.scatter(x, y, s=sizes, c=colors, alpha=0.7)
plt.title("Scatter Plot")
plt.show()


# ----- 3. RECTANGLE -----
fig, ax = plt.subplots()
rect = patches.Rectangle((1,1), 2, 3, linewidth=2, edgecolor='r', facecolor='yellow')
ax.add_patch(rect)
plt.xlim(0,5)
plt.ylim(0,5)
plt.title("Rectangle")
plt.show()


# ----- 4. CIRCLE -----
fig, ax = plt.subplots()
circle = patches.Circle((2,2), radius=1, edgecolor='b', facecolor='green', alpha=0.5)
ax.add_patch(circle)
plt.xlim(0,5)
plt.ylim(0,5)
plt.title("Circle")
plt.show()


# ----- 5. ELLIPSE -----
fig, ax = plt.subplots()
ellipse = patches.Ellipse((2,2), width=3, height=1.5, edgecolor='purple', facecolor='pink', alpha=0.5)
ax.add_patch(ellipse)
plt.xlim(0,5)
plt.ylim(0,5)
plt.title("Ellipse")
plt.show()


# ----- 6. POLYGON -----
fig, ax = plt.subplots()
polygon = patches.Polygon([[1,1],[2,3],[3,1],[2,0]], closed=True, edgecolor='black', facecolor='orange', alpha=0.5)
ax.add_patch(polygon)
plt.xlim(0,4)
plt.ylim(0,4)
plt.title("Polygon")
plt.show()


# ----- 7. BAR PLOT -----
x = ['A','B','C','D']
y = [5,7,3,8]
plt.bar(x, y, color='cyan')
plt.title("Bar Plot")
plt.show()


# ----- 8. HISTOGRAM -----
data = np.random.randn(1000)
plt.hist(data, bins=20, color='lightgreen', edgecolor='black')
plt.title("Histogram")
plt.show()


# ----- 9. PIE CHART -----
sizes = [15, 30, 45, 10]
labels = ['A','B','C','D']
colors = ['gold', 'lightblue', 'lightgreen', 'pink']
plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
plt.title("Pie Chart")
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Create figure and axes
fig, ax = plt.subplots(figsize=(8,8))

# ----- RECTANGLE -----
rect = patches.Rectangle((1,1), 2, 3, linewidth=2, edgecolor='red', facecolor='yellow', alpha=0.5)
ax.add_patch(rect)

# ----- CIRCLE -----
circle = patches.Circle((5,5), radius=1.5, edgecolor='blue', facecolor='green', alpha=0.5)
ax.add_patch(circle)

# ----- ELLIPSE -----
ellipse = patches.Ellipse((8,2), width=3, height=1.5, edgecolor='purple', facecolor='pink', alpha=0.5)
ax.add_patch(ellipse)

# ----- POLYGON -----
polygon = patches.Polygon([[6,7],[7,9],[9,8],[8,6]], closed=True, edgecolor='black', facecolor='orange', alpha=0.5)
ax.add_patch(polygon)

# Set limits and title
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.set_aspect('equal')
plt.title("All Common 2D Shapes")
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Create figure and axes
fig, ax = plt.subplots(figsize=(6,6))

# ----- RECTANGLE -----
rect = patches.Rectangle((1, 1), 3, 2, edgecolor='red',facecolor='yellow', alpha=0.5)
ax.add_patch(rect)

# ----- LINE -----
ax.plot([0, 5], [0, 5], color='blue', linewidth=2, label="Diagonal Line")

# Set limits and title
ax.set_xlim(0, 6)
ax.set_ylim(0, 6)
ax.set_aspect('equal')
plt.title("Rectangle and Line")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Create figure and axes
fig, ax = plt.subplots(figsize=(8,8))

# ----- RECTANGLE -----
rect = patches.Rectangle(
    (1, 1),     # bottom-left corner
    3, 2,       # width, height
    edgecolor='red',
    facecolor='yellow',
    linewidth=2,
    alpha=0.5
)
ax.add_patch(rect)

# ----- CIRCLE -----
circle = patches.Circle(
    (6, 6),     # center
    1.5,        # radius
    edgecolor='blue',
    facecolor='green',
    linewidth=2,
    alpha=0.5
)
ax.add_patch(circle)

# ----- LINE -----
ax.plot(
    [0, 8], [0, 8],   # start and end points
    color='black',
    linewidth=2,
    linestyle='--',
    marker='o'
)

# ----- POLYGON -----
polygon = patches.Polygon(
    [[2,7], [3,9], [5,8], [4,6]],  # vertices
    closed=True,
    edgecolor='purple',
    facecolor='orange',
    alpha=0.5
)
ax.add_patch(polygon)

# ----- SETTINGS -----
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.set_aspect('equal')
plt.title("Basic Shapes Control")
plt.show()
```
