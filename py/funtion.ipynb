```python
# ============================================
#        PYTHON FUNCTIONS — ALL IN ONE
# ============================================

# ----- 1. BASIC FUNCTION -----
def greet():
    print("Hello, World!")

greet()


# ----- 2. FUNCTION WITH PARAMETERS -----
def greet_person(name):
    print(f"Hello, {name}!")

greet_person("Bappy")


# ----- 3. FUNCTION WITH RETURN -----
def add(a, b):
    return a + b

result = add(5, 3)
print("Add:", result)


# ----- 4. DEFAULT PARAMETERS -----
def greet_with_msg(name="Guest"):
    print(f"Hello, {name}!")

greet_with_msg()        # uses default
greet_with_msg("Fariha") # overrides default


# ----- 5. KEYWORD ARGUMENTS -----
def person_info(name, age):
    print(f"Name: {name}, Age: {age}")

person_info(age=22, name="Bappy")  # keyword order can change


# ----- 6. VARIABLE-LENGTH ARGUMENTS -----
# *args → tuple, **kwargs → dictionary

def fruits(*args):
    print("Fruits:", args)

fruits("apple", "banana", "mango")

def details(**kwargs):
    print("Details:", kwargs)

details(name="Bappy", age=22)


# ----- 7. RECURSION -----
# Factorial example
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

print("Factorial 5:", factorial(5))


# ----- 8. LAMBDA FUNCTION (ANONYMOUS FUNCTION) -----
square = lambda x: x * x
print("Square 4:", square(4))

add_lambda = lambda a, b: a + b
print("Add lambda:", add_lambda(5, 3))


# ----- 9. FUNCTION INSIDE FUNCTION (NESTED FUNCTION) -----
def outer(x):
    def inner(y):
        return y * 2
    return inner(x) + 5

print("Nested:", outer(3))


# ----- 10. PASS FUNCTION AS ARGUMENT -----
def apply_func(func, value):
    return func(value)

print("Apply lambda:", apply_func(lambda x: x + 10, 5))


# ----- 11. DOCSTRING -----
def greet_doc():
    """This function prints Hello"""
    print("Hello")

print(greet_doc.__doc__)


# ----- 12. ANNOTATIONS -----
def add_annot(a: int, b: int) -> int:
    return a + b

print("Annotated add:", add_annot(2, 3))


# ============================================
#                   END
# ============================================
```

```python
def fib_single(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fib_single(n-1) + fib_single(n-2)

print("10th Fibonacci number:", fib_single(9))  # 0-based index

n = 10
print("Fibonacci sequence:")
for i in range(n):
    print(fib_single(i), end=" ")
```

```python
# ----- 5. USING DYNAMIC PROGRAMMING -----
def fibonacci_dp(n):
    if n <= 0:
        return []
    fib = [0, 1]
    for i in range(2, n):
        fib.append(fib[i-1] + fib[i-2])
    return fib[:n]

print("DP Method:", fibonacci_dp(10))
```

CLASS

```python
# ----- 1. BASIC CLASS -----
class Person:
    def __init__(self, name, age):   # constructor
        self.name = name
        self.age = age

    def greet(self):                 # method
        print(f"Hello, I am {self.name} and I am {self.age} years old.")

# Create object
p1 = Person("Bappy", 22)
p1.greet()


# ----- 2. INHERITANCE -----
class Student(Person):              # Student inherits Person
    def __init__(self, name, age, course):
        super().__init__(name, age)
        self.course = course

    def show_course(self):
        print(f"{self.name} studies {self.course}")

s1 = Student("Fariha", 20, "CSE")
s1.greet()         # inherited method
s1.show_course()   # own method


# ----- 3. CLASS & INSTANCE VARIABLES -----
class Example:
    class_var = "I am class variable"  # shared by all objects

    def __init__(self, value):
        self.instance_var = value       # unique to each object

e1 = Example(5)
e2 = Example(10)
print(e1.class_var, e1.instance_var)
print(e2.class_var, e2.instance_var)


# ----- 4. METHOD TYPES -----
class Demo:
    def instance_method(self):
        print("Instance method", self)

    @classmethod
    def class_method(cls):
        print("Class method", cls)

    @staticmethod
    def static_method():
        print("Static method")

d = Demo()
d.instance_method()
Demo.class_method()
Demo.static_method()

# ============================================
#                 END
# ============================================
```
