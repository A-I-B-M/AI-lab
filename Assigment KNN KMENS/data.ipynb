```python
# =====================================================
# EXTENDED DATASET MANIPULATION (EXAM FOCUS)
# No model logic — only data operations
# =====================================================

import pandas as pd
import numpy as np

# -----------------------------------------------------
# 1. DATASET WITH ISSUES (REAL EXAM TYPE)
# -----------------------------------------------------
data = pd.DataFrame({
    "Age": [22, 25, np.nan, 40, 30],
    "Salary": [20000, 30000, 60000, np.nan, 50000],
    "Experience": [1, 2, 5, 6, 4],
    "Purchased": ["No", "No", "Yes", "Yes", "Yes"]
})

# COMMENT:
# Dataset may contain missing values & categorical labels

# -----------------------------------------------------
# 2. VIEW DATASET (WRITE IN THEORY)
# -----------------------------------------------------
data.head()
data.info()

# COMMENT:
# Used to understand structure & missing values

# -----------------------------------------------------
# 3. COLUMN SELECTION (MULTIPLE WAYS)
# -----------------------------------------------------

# By name
X1 = data[["Age", "Salary"]]

# By index
X2 = data.iloc[:, 0:2]

# Mixed selection
X3 = data.loc[:, "Age":"Salary"]

# COMMENT:
# Feature selection changes distance computation

# -----------------------------------------------------
# 4. ROW SELECTION (VERY IMPORTANT)
# -----------------------------------------------------

# First N rows
rows1 = data.head(3)

# Last N rows
rows2 = data.tail(2)

# Specific row
row3 = data.iloc[2]

# Row range
rows4 = data.iloc[1:4]

# COMMENT:
# Row selection used for sampling or subset training

# -----------------------------------------------------
# 5. CONDITIONAL ROW SELECTION
# -----------------------------------------------------
high_salary = data[data["Salary"] > 40000]

# COMMENT:
# Filters dataset based on condition

# -----------------------------------------------------
# 6. HANDLE MISSING VALUES (NO DELETION)
# -----------------------------------------------------

# Mean filling
data["Age"].fillna(data["Age"].mean(), inplace=True)

# Median filling
data["Salary"].fillna(data["Salary"].median(), inplace=True)

# COMMENT:
# Mean/Median preserves dataset size

# -----------------------------------------------------
# 7. FORWARD / BACKWARD FILL
# -----------------------------------------------------
data.fillna(method="ffill", inplace=True)
data.fillna(method="bfill", inplace=True)

# COMMENT:
# Useful for time-series like data

# -----------------------------------------------------
# 8. CATEGORICAL → NUMERIC
# -----------------------------------------------------
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
data["Purchased"] = le.fit_transform(data["Purchased"])

# COMMENT:
# Distance models require numeric data

# -----------------------------------------------------
# 9. TYPE CONVERSION
# -----------------------------------------------------
data["Age"] = data["Age"].astype(int)

# COMMENT:
# Ensures consistent numeric computation

# -----------------------------------------------------
# 10. DATA NORMALIZATION (MIN-MAX)
# -----------------------------------------------------
from sklearn.preprocessing import MinMaxScaler
mm = MinMaxScaler()
data[["Age", "Salary"]] = mm.fit_transform(data[["Age", "Salary"]])

# COMMENT:
# Scales data between 0 and 1

# -----------------------------------------------------
# 11. STANDARDIZATION (Z-SCORE)
# -----------------------------------------------------
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
data[["Age", "Salary"]] = sc.fit_transform(data[["Age", "Salary"]])

# COMMENT:
# Mean = 0, Std = 1

# -----------------------------------------------------
# 12. RESHAPING DATA
# -----------------------------------------------------
X = data[["Age", "Salary"]].values
X_reshaped = X.reshape(-1, 2)

# COMMENT:
# Required for sklearn input format

# -----------------------------------------------------
# 13. RANDOM SAMPLING
# -----------------------------------------------------
sample_data = data.sample(frac=0.6, random_state=42)

# COMMENT:
# Used for random subset selection

# -----------------------------------------------------
# 14. SORTING DATA
# -----------------------------------------------------
sorted_data = data.sort_values(by="Salary")

# COMMENT:
# Useful for understanding distribution

# -----------------------------------------------------
# 15. DUPLICATE HANDLING (NO DELETE)
# -----------------------------------------------------
data["is_duplicate"] = data.duplicated()

# COMMENT:
# Identifies duplicate rows

# -----------------------------------------------------
# 16. ADD NEW FEATURE (FEATURE ENGINEERING)
# -----------------------------------------------------
data["Salary_per_Exp"] = data["Salary"] / (data["Experience"] + 1)

# COMMENT:
# New feature improves distance representation

# -----------------------------------------------------
# 17. SPLIT FEATURES & TARGET
# -----------------------------------------------------
X_final = data.drop(columns=["Purchased"])
y_final = data["Purchased"]

# COMMENT:
# Required before model training

# =====================================================
# END — ALL DATASET OPERATIONS COVERED
# =====================================================
```
